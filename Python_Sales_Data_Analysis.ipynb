```python
import pandas as pd

data = {
    "Product": ["Laptop", "Phone", "Mouse", "Keyboard"],
    "Price": [5000, 3000, 150, 250],
    "Quantity": [3, 5, 10, 4],
    "City": ["Lusaka", "Ndola", "Kitwe", "Livingstone"]
}

df = pd.DataFrame(data)

display(df)
```

```python
df["Total_Sales"] = df["Price"] * df["Quantity"]

display(df)
```

```python
print("Total Revenue: K", df["Total_Sales"].sum())

print("Average Sales: K", df["Total_Sales"].mean())

print("Highest Sale: K", df["Total_Sales"].max())

print("Lowest Sale: K", df["Total_Sales"].min())

print("Total Products:", len(df))
```

```python
lusaka_sales = df[df["City"] == "Lusaka"]

display(lusaka_sales)
```

```python
high_sales = df[df["Total_Sales"] > 5000]

display(high_sales)
```

```python
df[df["Total_Sales"] < 5000]
```

```python
sorted_sales = df.sort_values(
    by="Total_Sales",
    ascending=False
)

display(sorted_sales)
```

```python
sorted_sales = df.sort_values(
    by="Total_Sales",
    ascending=True
)

display(sorted_sales)
```

```python
df.sort_values(by="Price", ascending=False)
```

```python
sales_by_product = df.groupby("Product")["Total_Sales"].sum()

display(sales_by_product)
```

```python
sales_by_city = (
    df.groupby("City")["Total_Sales"]
    .sum()
    .sort_values(ascending=False)
)

display(sales_by_city)
```

```python
import matplotlib.pyplot as plt

sales_by_city = df.groupby("City")["Total_Sales"].sum()

plt.figure(figsize=(8, 5))

plt.bar(sales_by_city.index, sales_by_city.values)

plt.title("Total Sales by City")
plt.xlabel("City")
plt.ylabel("Total Sales (ZMW)")

plt.show()
```

```python
sales_by_city = (
    df_clean.groupby("City")["Total Sales"]
    .sum()
    .sort_values(ascending=False)
)

print("Sales Performance by City")
display(sales_by_city)
```

```python
import matplotlib.pyplot as plt

# Calculate sales by city
sales_by_city = (
    df_clean.groupby("City")["Total Sales"]
    .sum()
    .sort_values(ascending=False)
)

# Create the chart
plt.figure(figsize=(10, 6))

plt.bar(
    sales_by_city.index,
    sales_by_city.values,
    color="steelblue"
)

# Add chart labels
plt.title("Sales Performance by City", fontsize=16)
plt.xlabel("City")
plt.ylabel("Total Revenue (ZMW)")

# Show sales amounts above the bars
for i, value in enumerate(sales_by_city.values):
    plt.text(i, value + 300, f"K{value:,.0f}", ha="center")

plt.tight_layout()
plt.show()
```

```python
import pandas as pd

# Correct Excel file path
file_path = "/content/Sales_Performance_Dashboard.xlsx (1) (1).XLSX"

# Load sales data
df = pd.read_excel(file_path, sheet_name="SalesData")

# Remove summary rows
df_clean = df[
    df["Order ID"].astype(str).str.match(r"^ORD\d+$")
].copy()

# Verify the results
print("Number of Orders:", len(df_clean))
print("Total Revenue: K", df_clean["Total Sales"].sum())

display(df_clean.head())
```

```python
# Calculate total revenue by category
sales_by_category = (
    df_clean.groupby("Category")["Total Sales"]
    .sum()
    .sort_values(ascending=False)
)

print("Sales Performance by Category")

display(sales_by_category)
```

```python
import matplotlib.pyplot as plt

# Calculate total sales by category
sales_by_category = (
    df_clean.groupby("Category")["Total Sales"]
    .sum()
    .sort_values(ascending=False)
)

# Create bar chart
plt.figure(figsize=(9, 5))

bars = plt.bar(
    sales_by_category.index,
    sales_by_category.values,
    color="steelblue"
)

# Add title and labels
plt.title("Sales Performance by Category", fontsize=16)
plt.xlabel("Product Category")
plt.ylabel("Total Revenue (ZMW)")

# Add revenue values above bars
for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        height + 500,
        f"K{height:,.0f}",
        ha="center"
    )

plt.ylim(0, sales_by_category.max() * 1.15)
plt.tight_layout()
plt.show()
```

```python
# Calculate total revenue by salesperson
sales_by_person = (
    df_clean.groupby("Salesperson")["Total Sales"]
    .sum()
    .sort_values(ascending=False)
)

print("Sales Performance by Salesperson")

display(sales_by_person)
```

```python
import matplotlib.pyplot as plt

# Calculate total revenue by salesperson
sales_by_person = (
    df_clean.groupby("Salesperson")["Total Sales"]
    .sum()
    .sort_values(ascending=False)
)

# Create chart
plt.figure(figsize=(9, 5))

bars = plt.bar(
    sales_by_person.index,
    sales_by_person.values,
    color="steelblue"
)

# Add chart title and labels
plt.title("Sales Performance by Salesperson", fontsize=16)
plt.xlabel("Salesperson")
plt.ylabel("Total Revenue (ZMW)")

# Display revenue above each bar
for bar in bars:
    height = bar.get_height()

    plt.text(
        bar.get_x() + bar.get_width() / 2,
        height + 500,
        f"K{height:,.0f}",
        ha="center"
    )

plt.ylim(0, sales_by_person.max() * 1.15)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd

# Convert Date column to datetime
df_clean["Date"] = pd.to_datetime(df_clean["Date"])

# Calculate revenue by date
daily_sales = (
    df_clean.groupby("Date")["Total Sales"]
    .sum()
    .sort_index()
)

print("Daily Sales Performance")

display(daily_sales)
```

```python
import matplotlib.pyplot as plt

# Calculate daily sales revenue
daily_sales = (
    df_clean.groupby("Date")["Total Sales"]
    .sum()
    .sort_index()
)

# Create line chart
plt.figure(figsize=(12, 6))

plt.plot(
    daily_sales.index,
    daily_sales.values,
    marker="o",
    color="steelblue",
    linewidth=2
)

# Add title and labels
plt.title("Daily Sales Revenue Trend", fontsize=16)
plt.xlabel("Date")
plt.ylabel("Total Revenue (ZMW)")

# Improve readability
plt.xticks(rotation=45)
plt.grid(True, linestyle="--", alpha=0.4)

plt.tight_layout()
plt.show()
```
